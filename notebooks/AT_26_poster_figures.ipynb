```python
from cycler import cycler
cmap = 'tab20'
n_colors = 6
cmap = plt.get_cmap(cmap)
colors = [cmap(i) for i in np.linspace(0,1,n_colors)]
plt.rcParams["axes.prop_cycle"] = cycler(color = colors)
cmap
```

```python
x0 = 3.0                
x = np.linspace(0, 10, 1001)

y1 = (x >= x0).astype(float)
y2 = (x <= x0).astype(float)

fig, ax = plt.subplots(figsize=(10, 10), layout = "tight")
ax.step(x, y1, where="post", lw = 5, linestyle = "--")
ax.step(x, y2, where = "post", lw = 5, linestyle = "--")         
ax.set(xlabel="Running Speed", ylabel="Activity", ylim=(-0.1, 1.1))
# ax.set_title(f"Step function at x0={x0:g}")
ax.set_xticks([])
ax.set_yticks([])
plt.show()

```
